Load the synthetic IT service-request dataset.

```python
from pathlib import Path
import pandas as pd
DATA = Path('..') / 'datasets' / 'service_requests.csv'
df = pd.read_csv(DATA)
df
```

Prepare inputs and split training from test data.

```python
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import classification_report
work = df.copy()
work['service_down'] = work['service_down'].replace({'yes':1,'no':0})
X = work[['wait_minutes','affected_users','service_down']]
y = work['priority']
X_train, X_test, y_train, y_test = train_test_split(X,y,test_size=0.30,random_state=42,stratify=y)
model = DecisionTreeClassifier(max_depth=2, random_state=42)
model.fit(X_train,y_train)
print(classification_report(y_test, model.predict(X_test), zero_division=0))
```
