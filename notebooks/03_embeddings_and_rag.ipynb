Use local vector similarity to retrieve a relevant knowledge-base entry and build a grounded answer.

Load the miniature knowledge base.

```python
from pathlib import Path
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
DATA = Path('..') / 'datasets' / 'mini_knowledge_base.csv'
docs = pd.read_csv(DATA)
docs
```

Rank documents for a question.

```python
question = 'What should happen when equipment readings are abnormal?'
vec = TfidfVectorizer(stop_words='english')
X = vec.fit_transform(docs['text'])
q = vec.transform([question])
scores = cosine_similarity(q, X).flatten()
result = docs.assign(similarity=scores).sort_values('similarity', ascending=False)
result
```

Use the top source as context. A production RAG system would still evaluate source quality and generation quality.

```python
source = result.iloc[0]
print('Retrieved:', source['doc_id'], source['text'])
```
